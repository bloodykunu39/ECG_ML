```python
def normalize_matrix(matrix):
    """take input as 500*12 or a person report and return the normalized matrix 
    >>>sum of square of all the elements of the norm_matrix is 1"""
    
    matrix_norm = np.linalg.norm(matrix, 'fro')#frobenius norm is the square root of the sum of the absolute squares of its elements
                                                # works for 2d arrays
    if np.isnan(matrix_norm):# is NAN not a number 
        #it will return a matrix of zeros of the same shape as the input matrix
        return np.zeros_like(matrix)
    else:
        normalized_matrix = matrix / matrix_norm
        return normalized_matrix

def superposition(data: np.ndarray, n, poly_order=None) -> np.ndarray:
    """
    This function takes the data matrix and returns the superposition data matrix

    args:
    data : np.ndarray : data matrix of shape (5000, 12)
    n    : int        : size of the output (n x n) superposition matrix
    poly_order : list/array of int, optional
                 permutation (or any custom ordering) of Legendre polynomial
                 degrees to use for each column of `data`.
                 Must have the same length as data.shape[1].
                 If None, defaults to [1, 2, ..., data.shape[1]] (i.e. i+1),
                 same behavior as before.

    returns:
    np.ndarray : superpositioned img matrix of shape (n, n)
    """

    scale = StandardScaler()
    scaled = scale.fit_transform(normalize_matrix(data))

    # default = old behavior: column i -> legendre order i+1
    if poly_order is None:
        poly_order = [i + 1 for i in range(data.shape[1])]

    if len(poly_order) != data.shape[1]:
        raise ValueError(
            f"poly_order length ({len(poly_order)}) must match "
            f"number of columns in data ({data.shape[1]})"
        )

    img = np.zeros((n, n))
    x = np.linspace(-1, 1, n)

    for i, order in enumerate(poly_order):
        leg = legendre(order)(x)
        norm = math.sqrt(np.sum(leg * leg))
        img = img + np.outer(scaled[:, i], leg) / norm

    return img

def process_report(report, poly_order=None, cg=50, n=5000):
    return coarsegrain(superposition(report, n=n, poly_order=poly_order), cg=cg)
```

```python
import os
import numpy as np
from time import time
from joblib import Parallel, delayed


def permutation_typ1(
    data_dir="data_prep",
    name="data_leg_typ1_50",
    disease_types=("SR", "ST", "SB"),
    split_number=10,
    n_jobs=20,
    poly_order=None,
    cg=50,
    n=5000,
):
    """
    Loads ECG datasets for each disease type, splits each into batches,
    processes each report in parallel using process_report, and saves
    the flattened results.
    """

    save_dir = os.path.join("ML", str(name))
    os.makedirs(save_dir, exist_ok=True)

    all_saved_paths = {}

    for disease in disease_types:
        data_path = os.path.join(data_dir, f"unq_disease_{disease}.npy")
        print(f"\n=== Processing {disease} ({data_path}) ===")

        # Memory-mapped loading (optional)
        data = np.load(data_path, mmap_mode="r")

        if len(data) % split_number != 0:
            raise ValueError(
                f"[{disease}] len(data)={len(data)} is not evenly divisible "
                f"by split_number={split_number}."
            )

        data_split = np.array_split(data, split_number)
        saved_paths = []

        for i, cc in enumerate(data_split):
            start = time()

            results = Parallel(n_jobs=n_jobs)(
                delayed(process_report)(
                    report,
                    poly_order=poly_order,
                    cg=cg,
                    n=n,
                )
                for report in cc
            )

            elapsed = time() - start
            print(
                f"  Batch {i + 1}/{split_number} "
                f"({len(cc)} reports) processed in {elapsed:.2f} seconds"
            )

            # Convert list of matrices into 2D array
            flattened = np.asarray([mat.ravel() for mat in results])

            save_path = os.path.join(
                save_dir,
                f"unq_disease_{disease}_svd_{i}.txt",
            )

            np.savetxt(save_path, flattened)
            saved_paths.append(save_path)

        all_saved_paths[disease] = saved_paths

    return all_saved_paths
```
