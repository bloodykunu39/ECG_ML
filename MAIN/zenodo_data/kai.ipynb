```python
import os

MAX_DEPTH = 3
TARGET = "/home/karansingh/Documents/summer term/ECG_ML/MAIN"


def human_readable(size):
    for unit in ['B', 'K', 'M', 'G', 'T']:
        if size < 1024:
            return f"{size:.1f}{unit}"
        size /= 1024


def build_size_map(root):
    size_map = {}

    for dirpath, dirnames, filenames in os.walk(root):
        total = 0
        for f in filenames:
            try:
                fp = os.path.join(dirpath, f)
                total += os.path.getsize(fp)
            except:
                pass
        size_map[dirpath] = total

    # propagate sizes upward
    for path in sorted(size_map.keys(), key=len, reverse=True):
        parent = os.path.dirname(path)
        if parent in size_map:
            size_map[parent] += size_map[path]

    return size_map


def print_tree(path, size_map, depth=0):
    if depth > MAX_DEPTH:
        return

    indent = "│   " * depth
    print(f"{indent}├── {os.path.basename(path)} [{human_readable(size_map.get(path, 0))}]")

    try:
        entries = list(os.scandir(path))

        files = []
        folders = []

        for entry in entries:
            if entry.is_file():
                try:
                    files.append((entry.name, os.path.getsize(entry.path)))
                except:
                    pass
            elif entry.is_dir(follow_symlinks=False):
                folders.append((entry.path, size_map.get(entry.path, 0)))

        # 🔥 First depth: show files + folders sorted
        if depth == 0:
            files.sort(key=lambda x: x[1], reverse=True)
            folders.sort(key=lambda x: x[1], reverse=True)

            for name, size in files:
                print(f"{indent}│   ├── {name} [{human_readable(size)}]")

            for folder_path, size in folders:
                print_tree(folder_path, size_map, depth + 1)

        else:
            # deeper levels → only folders, sorted
            folders.sort(key=lambda x: x[1], reverse=True)
            for folder_path, size in folders:
                print_tree(folder_path, size_map, depth + 1)

    except (PermissionError, FileNotFoundError):
        pass


size_map = build_size_map(TARGET)
print_tree(TARGET, size_map)
```

```python
import os
from collections import defaultdict

TARGET = "/home/karansingh/Documents/summer term/ECG_ML/MAIN"


def human_readable(size):
    for unit in ['B', 'K', 'M', 'G', 'T']:
        if size < 1024:
            return f"{size:.1f}{unit}"
        size /= 1024


type_count = defaultdict(int)
type_size = defaultdict(int)

for root, dirs, files in os.walk(TARGET):
    for file in files:
        path = os.path.join(root, file)

        try:
            size = os.path.getsize(path)
        except:
            continue

        ext = os.path.splitext(file)[1].lower()
        if ext == "":
            ext = "NO_EXTENSION"

        type_count[ext] += 1
        type_size[ext] += size


# Sort by total size (largest first)
sorted_types = sorted(type_size.items(), key=lambda x: x[1], reverse=True)

print("\nFile Type Analysis (Full Depth)\n")
print(f"{'TYPE':15} {'COUNT':10} {'TOTAL SIZE'}")
print("-" * 45)

for ext, size in sorted_types:
    print(f"{ext:15} {type_count[ext]:10} {human_readable(size)}")
```
