# Saving data

```python
base_dir = os.getcwd()  # wherever you launched Jupyter from
save_dir = os.path.join(base_dir, "ML", "data_unq")
save_dir
```

```python
import numpy as np
from joblib import Parallel, delayed
from time import time
import os

from encoding import superposition
from smoothening import coarsegrain

# helper function for joblib
def process_report(report):
    return coarsegrain(superposition(report, n=5000), cg=50)

# -------------------
# Load ECG dataset
# -------------------
data = np.load('data_prep/unq_disease_SR.npy')   # shape (N, 5000, 12)
split_number = 10

if len(data) % split_number == 0:
    data_split = np.array_split(data, split_number)

    save_dir = os.path.join("ML", "data_unq")
    os.makedirs(save_dir, exist_ok=True)

    for i in range(split_number):
        cc = data_split[i]  # 500 reports

        start = time()
        results = Parallel(n_jobs=20)(
            delayed(process_report)(report) for report in cc
        )
        end = time()
        print(f"Batch {i} of size {len(cc)} processed in {end - start:.2f} seconds")

        # flatten each (100x100) → (10000,)
        flattened = [mat.ravel() for mat in results]

        # save
        save_path = os.path.join(save_dir, f"unq_disease_SR_svd_{i}.txt")
        np.savetxt(save_path, flattened)
```
